```python
import pandas as pd
import yfinance as yf
import pyarrow
from datetime import datetime, timedelta
```

```python
list_tickers = [
    "PETR4.SA",
    "VALE3.SA",
    "ITUB4.SA",
    "BBDC4.SA",
    "WEGE3.SA",
    "^BVSP",
]
day = str((datetime.now() - timedelta(days=2)).date())
list_df_tickers = list()
for ticker_name in list_tickers:
    ticker = yf.Ticker(ticker_name)
    df = ticker.history(start="2026-05-01", end="2026-05-01").reset_index()
    df["ticker"] = ticker_name
    list_df_tickers.append(df)
df = pd.concat(list_df_tickers)
df.reset_index(inplace=True)
try:
    if len(df[df["Date"] == "2026-04-30"]) > 0:
        raise Exception("Cu")
except Exception as e:
        print(f"Erro na carga: {e}")

```
